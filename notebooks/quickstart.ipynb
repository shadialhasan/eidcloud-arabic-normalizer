```python
# Cell 2: CLI Demonstration & JSON Output
# 1. Direct normalization with full diacritics stripping, Alef unification, and Tatweel removal
!php bin/eidcloud-ar-norm normalize "بِسْمِ اللَّـهِ الرَّحْمَٰنِ الرَّحِيمِ ... أَنَا أُحِبُّ اللُّغَـةَ العَـرَبِيَّـةَ"

# 2. Numeral standardization from Eastern (٠-٩) to Western (0-9) with JSON output format
!php bin/eidcloud-ar-norm normalize "سعر المنتج: ٤٥٠٠ ريال في عام ٢٠٢٦" --numerals=western --json

# 3. Diacritics extraction and position analysis
!php bin/eidcloud-ar-norm extract-tashkeel "مَرْحَبًا بِكُمْ" --json
```

```python
# Cell 3: High-Throughput Batch & File Streaming Cleaning
# Create sample raw dataset with diacritics, Tatweel, and mixed numerals
sample_data = """بِسْمِ اللَّـهِ الرَّحْمَٰنِ الرَّحِيمِ
الرَّقْمُ الوَطَنِيُّ: ٠١٢٣٤٥٦٧٨٩
مُـنَـظَّـمَـةُ البَـرْمَـجِـيَّـاتِ العَـرَبِـيَّـةِ
سَنَةُ التَّأْسِيسِ: ١٤٤٧هـ / ٢٠٢٦م"""

with open("raw_arabic.txt", "w", encoding="utf-8") as f:
    f.write(sample_data)

# Execute streaming cleaning on file with performance metrics
!php bin/eidcloud-ar-norm clean raw_arabic.txt --out=clean_arabic.txt --stats

# Inspect cleaned output
print("\n--- Cleaned Content ---")
with open("clean_arabic.txt", "r", encoding="utf-8") as f:
    print(f.read())
```
